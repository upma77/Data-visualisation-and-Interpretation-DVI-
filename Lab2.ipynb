```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# pivot table 
flights_pivot = flights.pivot(
    index="month",
    columns="year",
    values="passengers"
)

print(flights_pivot)
```

```python
# HeatMap
plt.figure(figsize=(14, 8))

sns.heatmap(
    flights_pivot,
    annot=True,
    fmt="d",
    cmap="YlGnBu"
)

plt.title("Monthly Passengers Heatmap")
plt.xlabel("Year")
plt.ylabel("Month")

plt.show()
```

```python
#Cluster map
sns.clustermap(
    flights_pivot,
    annot=True,
    fmt="d",
    cmap="YlGnBu",
    figsize=(14, 8)
)

plt.show()
```

```python
#LMplot
sns.lmplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.title("LM Plot: Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

plt.show()
```

```python
#PairGrid
g = sns.PairGrid(iris, hue="species")

g.map(sns.scatterplot)

g.add_legend()

plt.show()
```

Interpretation

The PairGrid shows pairwise relationships between the numerical columns of the iris dataset. Different colors represent different species, making it easier to compare the species and identify their patterns.

```python
#FacetGrid
g = sns.FacetGrid(
    tips,
    col="time",
    row="sex"
)

g.map_dataframe(
    sns.scatterplot,
    x="total_bill",
    y="tip"
)

g.set_axis_labels("Total Bill", "Tip")
g.set_titles("{row_name} - {col_name}")

plt.show()
```

Interpretation

The FacetGrid creates separate scatter plots for different combinations of gender and meal time. This allows us to compare the relationship between total bill and tip across different groups.

```python
#jointGrid
g = sns.JointGrid(
    data=tips,
    x="total_bill",
    y="tip"
)

g.plot(
    sns.scatterplot,
    sns.histplot
)

plt.show()
```

Interpretation

The JointGrid shows the relationship between total bill and tip in the center, while the histograms on the sides show the individual distributions of total bill and tip

### 2. Experiment with Different Seaborn Styles

```python
#Darkgrid Style
sns.set_style("darkgrid")

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.title("Darkgrid Style")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

plt.show()
```

Interpretation

The darkgrid style adds a dark background grid to the plot, which makes it easier to compare values visually.

```python
#Whitegrid Style
sns.set_style("whitegrid")

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.title("Whitegrid Style")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

plt.show()
```

Interpretation

The whitegrid style provides a clean white background with grid lines. It improves readability and is commonly used for statistical plots

```python
#Ticks Style
sns.set_style("ticks")

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.title("Ticks Style")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

sns.despine()

plt.show()
```

Interpretation

The ticks style provides a simple and clean appearance with axis ticks and minimal grid lines. It is useful when a less cluttered visualization is required.

### 3. Apply Various Color Palettes

```python
Set@ Palette
sns.set_palette("Set2")
plt.figure(figsize=(8, 5))

sns.barplot(
    data=tips,
    x="day",
    y="tip"
)

plt.title("Average Tip by Day - Set2 Palette")
plt.xlabel("Day")
plt.ylabel("Average Tip")

plt.show()
```

Interpretation

The Set2 palette provides soft and distinct colors, making the categories easy to differentiate.

```python
#Deep Palette
sns.set_palette("deep")
plt.figure(figsize=(8, 5))

sns.barplot(
    data=tips,
    x="day",
    y="tip"
)

plt.title("Average Tip by Day - Deep Palette")
plt.xlabel("Day")
plt.ylabel("Average Tip")

plt.show()
```

Interpretation

The deep palette provides relatively strong and visually distinct colors for categorical data.

```python
#Pastel Palette
sns.set_palette("pastel")
plt.figure(figsize=(8, 5))

sns.countplot(
    data=tips,
    x="day"
)

plt.title("Number of Customers by Day - Pastel Palette")
plt.xlabel("Day")
plt.ylabel("Number of Customers")

plt.show()
```

Interpretation

The pastel palette uses lighter colors and gives the visualization a soft and visually appealing appearance.

```python
#Bright Paletter
sns.set_palette("bright")

plt.figure(figsize=(8, 5))

sns.countplot(
    data=tips,
    x="day"
)

plt.title("Number of Customers by Day - Bright Palette")
plt.xlabel("Day")
plt.ylabel("Number of Customers")

plt.show()
```

Interpretation:
The bright palette uses vivid colors, which makes different categories highly distinguishable.

## 4. Add Appropriate Titles, Axis Labels, and Legends

```python
sns.set_style("whitegrid")
sns.set_palette("Set2")

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="time"
)

plt.title("Relationship Between Total Bill and Tip")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.legend(
    title="Gender / Time",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.show()
```

The scatter plot shows the relationship between total bill and tip. The title describes the purpose of the visualization, axis labels identify the variables, and the legend explains the different categories represented by color and marker style.
