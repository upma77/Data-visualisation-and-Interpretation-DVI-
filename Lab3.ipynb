# SECTION 1 — Introduction to Plotly Express

### Experiment 1: Scatter Plot

```python
df_2007 = df[df["year"] == 2007]

fig = px.scatter(
    df_2007,
    x="gdpPercap",
    y="lifeExp",
    title="GDP Per Capita vs Life Expectancy"
)

fig.show()
```

Interpretation

The graph generally shows that countries with higher GDP per capita tend to have higher life expectancy.

### Experiment 2: Bar Plot

```python
asia = df[(df["continent"] == "Asia") & (df["year"] == 2007)]

fig = px.bar(
    asia,
    x="country",
    y="pop",
    title="Population of Asian Countries"
)

fig.show()
```

### Experiment 3: Line Plot

```python
india = df[df["country"] == "India"]

fig = px.line(
    india,
    x="year",
    y="lifeExp",
    markers=True,
    title="Life Expectancy of India Over Time"
)

fig.show()
```

### Experiment 5: Box Plot

```python
fig = px.box(
    df_2007,
    x="continent",
    y="lifeExp",
    title="Life Expectancy by Continent"
)

fig.show()
```

### Experiment 6: Area Chart

```python
countries = df[
    df["country"].isin(["India", "China", "United States", "Japan"])
]

fig = px.area(
    countries,
    x="year",
    y="pop",
    color="country",
    title="Population Growth Over Time"
)

fig.show()
```

# SECTION 2 — Interactive Visualizations

```python
fig = px.scatter(
    df_2007,
    x="gdpPercap",
    y="lifeExp",
    color="continent",
    hover_name="country",
    hover_data=["pop"],
    title="GDP vs Life Expectancy"
)

fig.show()
```

### Experiment 9: Color and Size

```python
fig = px.scatter(
    df_2007,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="continent",
    hover_name="country",
    size_max=60,
    title="GDP vs Life Expectancy Bubble Plot"
)

fig.show()
```

### Experiment 10: Animation

```python
fig = px.scatter(
    df,
    x="gdpPercap",
    y="lifeExp",
    animation_frame="year",
    animation_group="country",
    size="pop",
    color="continent",
    hover_name="country",
    log_x=True,
    size_max=55,
    range_x=[100, 100000],
    range_y=[25, 90],
    title="Animated GDP vs Life Expectancy"
)

fig.show()
```

Interpretation

The animation allows us to observe how GDP, population, and life expectancy changed over time rather than looking at only one year.

# SECTION 3 — Geographical Visualizations

### Experiment 11: World Power Consumption Choropleth

```python
power = pd.read_csv("2014_World_Power_Consumption.csv")

power.head()
```

```python
power.columns
```

```python
data = dict(
    type="choropleth",

    locations=power["Country"],
    locationmode="country names",

    z=power["Power Consumption KWH"],

    text=power["Country"],

    colorscale="Viridis",

    colorbar={
        "title": "Power Consumption"
    }
)
```

```python
layout = dict(
    title="2014 World Power Consumption",

    geo=dict(
        showframe=False,
        projection={
            "type": "natural earth"
        }
    )
)
```

```python
fig = go.Figure(
    data=[data],
    layout=layout
)

fig.show()
```

Interpretation

Darker or more intense colors represent countries with higher power consumption. The geographic visualization makes regional patterns easier to identify than a normal table.

### Experiment 12: Election Data — USA Choropleth

```python
import pandas as pd
import plotly.graph_objects as go

# Load dataset
election = pd.read_csv("2012_Election_Data.csv")

# Remove extra spaces from column names
election.columns = election.columns.str.strip()

# Check column names
print(election.columns.tolist())
```

```python
import plotly.graph_objects as go

fig = go.Figure(
    data=go.Choropleth(
        locations=election["State Abv"],
        z=election["Voting-Age Population (VAP)"],
        locationmode="USA-states",
        colorscale="Blues",
        text=election["State"],
        colorbar_title="VAP"
    )
)

fig.update_layout(
    title="2012 Voting-Age Population by US State",
    geo_scope="usa"
)

fig.show()
```

### Interpretation

The choropleth makes it easy to compare voting-age population geographically across U.S. states.
